# AI-Based Business Location Selector: Exploratory Data Analysis (EDA)
**Domain:** Machine Learning, Geospatial Analytics, and Decision Support  
**Objective:** Explore distributions, detect collinearity, and visualize commercial variables across Bhopal candidate locations.

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (10, 5)
plt.rcParams["font.sans-serif"] = "DejaVu Sans"

# Load dataset
data_path = Path("../data/business_location_synthetic_dataset.csv")
if not data_path.exists():
    data_path = Path("business_location_synthetic_dataset.csv")
df = pd.read_csv(data_path)
print(f"Dataset shape: {df.shape}")
df.head()
```

## 1. Summary Statistics & Data Hygiene

```python
print(df.info())
print("\nMissing values per column:")
print(df.isnull().sum())
df.describe().T
```

## 2. Business Type Distribution & Rental Distributions

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.countplot(data=df, x="business_type", ax=axes[0], palette="Set2")
axes[0].set_title("Distribution of Candidate Business Types")
axes[0].set_xlabel("Business Category")
axes[0].set_ylabel("Record Count")

sns.boxplot(data=df, x="business_type", y="monthly_rent", ax=axes[1], palette="Set2")
axes[1].set_title("Monthly Rent Distribution by Business Type")
axes[1].set_xlabel("Business Category")
axes[1].set_ylabel("Monthly Rent (INR)")
plt.tight_layout()
plt.show()
```

## 3. Footfall vs. Target Demand Relationship

```python
plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=df,
    x="daily_footfall",
    y="target_demand",
    hue="income_level",
    style="business_type",
    alpha=0.7,
    palette="viridis"
)
plt.title("Pedestrian Footfall vs. Commercial Customer Demand")
plt.xlabel("Daily Pedestrian Footfall")
plt.ylabel("Observed / Target Daily Demand")
plt.show()
```

## 4. Correlation Matrix of Numerical Features

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns
corr = df[numeric_cols].corr()

plt.figure(figsize=(11, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", square=True, linewidths=0.5)
plt.title("Correlation Heatmap of Urban Location Features")
plt.show()
```
